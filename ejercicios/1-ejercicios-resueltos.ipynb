## Ejercicio 1. NumPy: arreglos, indexación y operaciones

Crea dos arreglos y realiza las siguientes tareas:
1. Genera un arreglo de 10 elementos con valores del 0 al 9.
2. Crea un arreglo bidimensional de forma 3x3 con valores aleatorios.
3. Selecciona la segunda fila y la tercera columna.
4. Calcula la suma, la media y el máximo de cada arreglo.
5. Multiplica elemento a elemento ambos arreglos (si tienen misma forma).

Pista: usa `np.arange`, `np.random.rand`, slicing e indexado.

```python
# Ejercicio 1
# 1) Arreglo de 10 elementos del 0 al 9
arr = np.arange(10)
print("arr =", arr)

# 2) Arreglo 3x3 con valores aleatorios
rng = np.random.default_rng(42)
b = rng.random((3, 3))
print("\nb =\n", b)

# 3) Segunda fila y tercera columna
segunda_fila = b[1, :]
tercera_columna = b[:, 2]
print("\nSegunda fila:", segunda_fila)
print("Tercera columna:", tercera_columna)

# 4) Suma, media y máximo de cada arreglo
print("\nSuma de arr =", arr.sum())
print("Media de arr =", arr.mean())
print("Máximo de arr =", arr.max())
print("Suma de b =", b.sum())
print("Media de b =", b.mean())
print("Máximo de b =", b.max())

# 5) Multiplicación elemento a elemento
arr_3x3 = np.arange(9).reshape(3, 3)
print("\narr_3x3 =\n", arr_3x3)
print("\nMultiplicación elemento a elemento:\n", b * arr_3x3)
```

## Ejercicio 4. Pandas: agrupación y agregación

Usa este DataFrame de ventas para responder:

- Cuánto se vendió por cada región?
- Cuál fue el total de ventas por producto?
- Cuál es el promedio de ventas por región?

`df.groupby(...)` y `agg(...)` serán útiles.

```python
# Ejercicio 4
ventas = pd.DataFrame({
    "region": ["Norte", "Sur", "Norte", "Sur", "Centro", "Norte"],
    "producto": ["A", "B", "A", "A", "B", "C"],
    "unidades": [10, 15, 20, 12, 18, 5],
    "precio": [100, 80, 100, 90, 75, 110]
})
ventas["total"] = ventas["unidades"] * ventas["precio"]
print(ventas)

# Ventas totales por región
print("\nVentas por región:")
print(ventas.groupby("region")["total"].sum())

# Total por producto
print("\nTotal por producto:")
print(ventas.groupby("producto")["total"].sum())

# Promedio de ventas por región
print("\nPromedio de total por región:")
print(ventas.groupby("region")["total"].mean())
```

## Ejercicio 5. Pandas: merge, join y concatenación

Crea dos DataFrames y resuelve las siguientes preguntas:
1. Concatenar verticalmente dos DataFrames con columnas iguales.
2. Hacer un merge por la columna `id` entre dos tablas.
3. Realizar un join por índice entre dos DataFrames.

Te recomendamos usar `pd.concat`, `pd.merge` y `df.join`.

```python
# Ejercicio 5
# Concatenación y unión de DataFrames

# ============================================================
# 5.1 Concatenación vertical
# ============================================================
# concat() permite apilar DataFrames uno debajo del otro.

clientes1 = pd.DataFrame({
    "id": [1, 2, 3],
    "valor": ["A", "B", "C"]
})

clientes2 = pd.DataFrame({
    "id": [4, 5, 6],
    "valor": ["D", "E", "F"]
})

concat_vertical = pd.concat(
    [clientes1, clientes2],
    ignore_index=True
)

print("Concatenación vertical:")
print(concat_vertical)


# ============================================================
# 5.2 Concatenación horizontal
# ============================================================

df1 = pd.DataFrame({
    "id": [1, 2, 3],
    "nombre": ["Ana", "Luis", "Marta"]
})

df2 = pd.DataFrame({
    "edad": [25, 30, 28],
    "ciudad": ["Ensenada", "Tijuana", "Mexicali"]
})

concat_horizontal = pd.concat(
    [df1, df2],
    axis=1
)
print("\nConcatenación horizontal:")
print(concat_horizontal)


# ============================================================
# 5.3 Merge: unión basada en una columna
# ============================================================
# merge() permite relacionar DataFrames mediante una columna común.
# En este caso, utilizamos "id" como clave de unión.

clientes = pd.DataFrame({
    "id": [1, 2, 3, 4],
    "nombre": ["Ana", "Luis", "Marta", "Carlos"]
})

ventas = pd.DataFrame({
    "id": [1, 2, 1, 3],
    "monto": [150, 200, 320, 400]
})

merge_df = pd.merge(
    clientes,
    ventas,
    on="id",
    how="inner"
)

print("\nMerge por columna 'id':")
print(merge_df)

merge_df = pd.merge(
    clientes,
    ventas,
    on="id",
    how="left"
)

print("\nMerge a la izquierda por columna 'id':")
print(merge_df)


# ============================================================
# 5.3 Join: unión basada en el índice
# ============================================================
# join() está diseñado principalmente para unir DataFrames
# utilizando sus índices.

df_izq = pd.DataFrame(
    {"valor_izq": [10, 20, 30]},
    index=[1, 2, 3]
)

df_der = pd.DataFrame(
    {"valor_der": [100, 200, 300]},
    index=[1, 2, 4]
)

print("\nDataFrame izquierdo:")
print(df_izq)

print("\nDataFrame derecho:")
print(df_der)


# ------------------------------------------------------------
# Join izquierdo
# ------------------------------------------------------------

join_df = df_izq.join(
    df_der,
    how="left"
)

print("\nJoin izquierdo:")
print(join_df)


# ------------------------------------------------------------
# Join derecho
# ------------------------------------------------------------

join_df = df_izq.join(
    df_der,
    how="right"
)

print("\nJoin derecho:")
print(join_df)


# ------------------------------------------------------------
# Join externo
# ------------------------------------------------------------

join_df = df_izq.join(
    df_der,
    how="outer"
)

print("\nJoin externo:")
print(join_df)


# ============================================================
# RESUMEN
# ============================================================
#
# concat()  -> apilar o concatenar DataFrames
# merge()   -> unir mediante columnas
# join()    -> unir mediante índices
#
# Regla práctica:
#
# ¿Quiero apilar DataFrames?
#     -> concat()
#
# ¿Quiero unir usando una COLUMNA?
#     -> merge()
#
# ¿Quiero unir usando el ÍNDICE?
#     -> join()
```

## Ejercicio 8. Álgebra de datos: vectores

Dados dos vectores, realiza:
1. Producto punto (`np.dot`).
2. Norma euclidiana (`np.linalg.norm`).
3. Suma y resta vectorial.
4. Compara el resultado del producto punto con la suma de productos elemento a elemento.

Ejemplo: `u = np.array([1,2,3])`, `v = np.array([4,5,6])`.

```python
# Ejercicio 8
u = np.array([1, 2, 3])
v = np.array([4, 5, 6])

producto_punto = np.dot(u, v)
print("Producto punto =", producto_punto)
print("Norma u =", np.linalg.norm(u))
print("Norma v =", np.linalg.norm(v))
print("Suma =", u + v)
print("Resta =", u - v)
print("Suma de productos elemento a elemento =", np.sum(u * v))
print("¿Coinciden? ", np.allclose(producto_punto, np.sum(u * v)))
```

## Ejercicio 9. Álgebra de datos: matrices

1. Crea dos matrices 2x2.
2. Calcula la suma y el producto matricial.
3. Calcula el determinante y la inversa de una de las matrices.
4. Verifica si la matriz es invertible.

Pista: usa `np.linalg.det`, `np.linalg.inv` y `np.dot`.

```python
# Ejercicio 9
A = np.array([[2, 1], [1, 3]])
B = np.array([[1, 0], [0, 1]])

print("Matriz A =\n", A)
print("\nMatriz B =\n", B)
print("A + B =\n", A + B)
print("\nA @ B =\n", A @ B)
print("\nDeterminante de A =", np.linalg.det(A))
print("\nInversa de A =\n", np.linalg.inv(A))

es_invertible = np.linalg.det(A) != 0
print("\n¿A es invertible? ", es_invertible)
```

## Desafío final

Construye un mini análisis sencillo con un DataFrame de estudiantes y calcula:
- promedio de calificaciones por grupo,
- número de estudiantes aprobados,
- una gráfica que muestre el rendimiento por grupo.

Utiliza todo lo visto en la sesión: NumPy, Pandas y Matplotlib.

```python
# Desafío final
estudiantes = pd.DataFrame({
    "grupo": ["A", "A", "B", "B", "C", "C"],
    "estudiante": ["Ana", "Luis", "Maria", "Jose", "Sofia", "Pedro"],
    "calificacion": [85, 72, 68, 91, 78, 80]
})

print("DataFrame de estudiantes:")
print(estudiantes)

promedio_por_grupo = estudiantes.groupby("grupo")["calificacion"].mean()
print("\nPromedio por grupo:")
print(promedio_por_grupo)

aprobados = estudiantes[estudiantes["calificacion"] >= 70].shape[0]
print("\nNúmero de estudiantes aprobados: ", aprobados)

plt.figure(figsize=(8, 5))
promedio_por_grupo.plot(kind="bar", color="darkgreen")
plt.title("Rendimiento promedio por grupo")
plt.xlabel("Grupo")
plt.ylabel("Calificación promedio")
plt.grid(axis="y", alpha=0.3)
plt.show()
```

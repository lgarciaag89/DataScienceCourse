````python
# generame una funcion que lea una cadenade aminoacidos y me devuelva los grupos funcionales presentes en la cadena de aminoácidos. Aquí tienes un ejemplo de cómo podrías implementar esta función en Python:```python
def identificar_grupos_funcionales(cadena_aminoacidos):
    # Diccionario de grupos funcionales y sus aminoácidos correspondientes
    grupos_funcionales = {
        'Amino': ['Gly', 'Ala', 'Val', 'Leu', 'Ile', 'Met', 'Phe', 'Trp', 'Tyr'],
        'Carboxilo': ['Asp', 'Glu'],
        'Hidroxilo': ['Ser', 'Thr'],
        'Sulfuro': ['Cys', 'Met'],
        'Amida': ['Asn', 'Gln'],
        'Básico': ['Lys', 'Arg', 'His'],
        'Ácido': ['Asp', 'Glu']
    }

    # Lista para almacenar los grupos funcionales presentes
    grupos_presentes = set()

    # Iterar sobre cada aminoácido en la cadena
    for aminoacido in cadena_aminoacidos.split('-'):
        for grupo, aminoacidos in grupos_funcionales.items():
            if aminoacido in aminoacidos:
                grupos_presentes.add(grupo)

    return list(grupos_presentes)
````

````python
identificar_grupos_funcionales('Gly-Ala-Val-Leu-Ile-Met-Phe-Trp-Tyr-Ser-Thr-Cys-Asn-Gln-Lys-Arg-His-Asp-Glu')
````

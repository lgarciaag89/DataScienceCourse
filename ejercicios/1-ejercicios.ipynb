## Ejercicio 7. Visualización con Matplotlib

Genera dos gráficas con los mismos datos:
1. Un gráfico de líneas para mostrar la evolución de una serie.
2. Un gráfico de barras para comparar categorías.
3. Añade título, etiquetas y una cuadrícula.

Ejemplo de datos: meses del año y ventas mensuales.

```python
# Ejercicio 7
meses = ['Ene', 'Feb', 'Mar', 'Abr', 'May', 'Jun']
ventas = [120, 150, 130, 180, 210, 200]

plt.figure(figsize=(10, 4))
plt.plot(meses, ventas, marker='o')
plt.title('Ventas mensuales')
plt.xlabel('Mes')
plt.ylabel('Ventas')
plt.grid(True)
plt.show()

plt.figure(figsize=(10, 4))
plt.bar(meses, ventas, color='steelblue')
plt.title('Comparación de ventas por mes')
plt.xlabel('Mes')
plt.ylabel('Ventas')
plt.grid(axis='y', alpha=0.3)
plt.show()
```

## Ejercicio 8. Álgebra de datos: vectores

Dados dos vectores, realiza:
1. Producto punto (`np.dot`).
2. Norma euclidiana (`np.linalg.norm`).
3. Suma y resta vectorial.
4. Compara el resultado del producto punto con la suma de productos elemento a elemento.

Ejemplo: `u = np.array([1,2,3])`, `v = np.array([4,5,6])`.

```python
# Ejercicio 8
u = np.array([1, 2, 3])
v = np.array([4, 5, 6])

print('Producto punto =', np.dot(u, v))
print('Norma u =', np.linalg.norm(u))
print('Norma v =', np.linalg.norm(v))
print('Suma =', u + v)
print('Resta =', u - v)
```

## Ejercicio 9. Álgebra de datos: matrices

1. Crea dos matrices 2x2.
2. Calcula la suma y el producto matricial.
3. Calcula el determinante y la inversa de una de las matrices.
4. Verifica si la matriz es invertible.

Pista: usa `np.linalg.det`, `np.linalg.inv` y `np.dot`.

```python
# Ejercicio 9
A = np.array([[2, 1], [1, 3]])
B = np.array([[1, 0], [0, 1]])

print('A + B = ', A + B)
print('A @ B = ', A @ B)
print('det(A) =', np.linalg.det(A))
print('inv(A) = ', np.linalg.inv(A))
```
